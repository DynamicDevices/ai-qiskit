## 🎮 Interactive Quantum Coin Flip

Let's explore the difference between classical and quantum coins! This demonstrates one of the most fundamental concepts in quantum mechanics.


```python
# 🪙 Classical Deterministic Coin
print("🪙 Classical Coin Flip (X gate)")
print("This always gives the same result - completely predictable!")

# Create a classical coin circuit
classical_coin = QuantumCircuit(1, 1)
classical_coin.x(0)  # Flip to |1⟩ state (Heads)
classical_coin.measure(0, 0)

print("\nCircuit:")
print(classical_coin.draw())

# Execute it multiple times
executor = QuantumExecutor()
classical_results = []

for i in range(5):
    result = executor.execute_circuit(classical_coin, shots=1)
    if result['success']:
        outcome = list(result['counts'].keys())[0]
        classical_results.append(int(outcome))
        print(f"Flip {i+1}: {'🟡 Heads' if outcome == '1' else '🔵 Tails'}")

print(f"\n📊 Results: Heads={classical_results.count(1)}/5, Tails={classical_results.count(0)}/5")
print("🎪 Always the same - this is deterministic!")
```

```python
# 🌊 Quantum Superposition Coin
print("🌊 Quantum Coin Flip (Hadamard gate)")
print("This creates TRUE randomness from quantum superposition!")

# Create a quantum coin in superposition
quantum_coin = QuantumCircuit(1, 1)
quantum_coin.h(0)  # Create superposition |0⟩ + |1⟩
quantum_coin.measure(0, 0)

print("\nCircuit:")
print(quantum_coin.draw())

# Execute it multiple times
quantum_results = []

for i in range(5):
    result = executor.execute_circuit(quantum_coin, shots=1)
    if result['success']:
        outcome = list(result['counts'].keys())[0]
        quantum_results.append(int(outcome))
        print(f"Flip {i+1}: {'🟡 Heads' if outcome == '1' else '🔵 Tails'}")

print(f"\n📊 Results: Heads={quantum_results.count(1)}/5, Tails={quantum_results.count(0)}/5")
print("🌊 Random outcomes - this is quantum mechanics!")
```

```python
# 🎯 Quantum Biased Coin - Adjustable Probability!
print("🎯 Quantum Biased Coin (Rotation gate)")
print("We can control the probability of outcomes!")

# Create a biased quantum coin
angle = np.pi / 3  # 60 degrees
biased_coin = QuantumCircuit(1, 1)
biased_coin.ry(angle, 0)  # Rotate the qubit state
biased_coin.measure(0, 0)

print(f"\nRotation angle: {angle:.2f} radians ({np.degrees(angle):.1f}°)")
print("Circuit:")
print(biased_coin.draw())

# Calculate theoretical probabilities
prob_tails = np.cos(angle/2)**2
prob_heads = np.sin(angle/2)**2
print(f"\n🧮 Theory predicts:")
print(f"  P(Tails) = {prob_tails:.3f}")
print(f"  P(Heads) = {prob_heads:.3f}")

# Test with many flips
result = executor.execute_circuit(biased_coin, shots=1000)
if result['success']:
    analysis = analyze_results(result)
    
    print(f"\n📊 Experimental results (1000 flips):")
    for state, prob in analysis['probabilities'].items():
        outcome = "Tails" if state == "0" else "Heads"
        print(f"  {outcome}: {prob:.3f}")
    
    # Show histogram
    plot_histogram(result['counts'], title='Biased Quantum Coin Results')
    plt.show()
    
    print("\n💡 By rotating the qubit, we can create any probability we want!")
```
